```python
import random

def criar_matriz(QT_LINHAS=4, QT_COLUNAS=4):
    matriz = []
    for linha in range (QT_LINHAS):
        linha=[]
        for coluna in range (QT_COLUNAS):
            linha.append(random.randint(1, 100))
        matriz.append(linha)
    return (matriz)

matriz = criar_matriz (6, 6)
print("Matriz:")
for linha in matriz:
    print(linha)

print("\nMaior Valor de cada linha:")
for i in range(4):
    maior_elemento=max(matriz[i])
    print(f"Maior valor da linha {i}: {maior_elemento}")
```

```python
linha_atual=1
for linha in minha_matriz:
    maior_numero=0
    for elemento in linha :
        if elemento > maior_numero:
            maior_numero = elemento
    print (f"O maior valor da {linha_atual}a. é {maior_numero}")
    linha_atual+=1
```

```python
#fromkeys
chaves = ["nome", "idade", "cidade"]
valor_padrao = "N/A"
usuario = dict.fromkeys(chaves, valor_padrao)

usuario["nome"] = "Hugo"
usuario["idade"]=22
usuario["cidade"]="Petrópolis"


print(usuario)
```
