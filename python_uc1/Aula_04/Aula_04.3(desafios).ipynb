## UTILIZANDO DESVIO CONDICIONAL

```python
numero=int(input("Digite o numero"))
if numero <0:
    print("O número é negativo")
elif numero ==0:
    print("O número é 0")
else:
    print("O número é positivo")
```

4️⃣ Faça um programa que peça um número e diga se ele é par ou ímpar.

```python
num=float(input("Digite o numero"))
if num %2==0:
    print("O numero é par")
elif num %2!=0:
    print("O numero é impar")
```

```python
peso=float(input("Digite seu peso em KG"))
altura=float(input("Digite sua altura em M"))
```

```python
imc=(peso/altura**2)
if imc<18.5:
    print("Você está abaixo do peso")
elif imc<=18.5:
    print("Você está no peso ideal")
else:
        print("Você está sobrepeso")
```

```python
valor_compra=float(input("Digite o valor total da compra"))
if valor_compra<200:
    percentual=0.05
elif valor_compra <=500:
    percentual=0.10
else:
    percentual=0.15
    
```

```python
valor_final=(valor_compra-(valor_compra*percentual))
print(f" O seu valor com final com desconto é: {valor_final}")
```
