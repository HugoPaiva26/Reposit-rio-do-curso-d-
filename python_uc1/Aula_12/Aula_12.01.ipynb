# Objetos

```python
class Carro:
    # Método construtor (__init__) para inicializar os atributos
    def __init__(self, marca, modelo, portas=4, ano="2026", cor = "Branca"):
        self.marca = marca      # Atributo
        self.modelo = modelo    # Atributo
        self.ano = ano          # Atributo
        self.cor=cor
        self.potencia=1.0
        self.tp_cambio="auto"
        self.qt_portas=portas
        self.dir_hidraulica=True
        self.tp_tracao=2
        self._cambio_automatico=False
        self._4p4=False
        self._ligado = False     # Atributo com valor padrão
        
      

    # Método para ligar o carro
    def ligar(self):
        if not self._ligado:
            self._ligado = True
            print(f"{self.marca} {self.modelo} está ligado.")
        else:
            print(f"{self.marca} {self.modelo} já está ligado.")

    # Método para desligar o carro
    def desligar(self):
        if self._ligado:
            self._ligado = False
            print(f"{self.marca} {self.modelo} está desligado.")
        else:
            print(f"{self.marca} {self.modelo} já está desligado.")

    # Método para exibir informações do carro
    def exibir_info(self):
        status = "ligado" if self._ligado else "desligado"
        print(f"{self.marca} {self.modelo} ({self.ano}) está {status}.")

    #Alterar cambio para manual
    def alterar_para_manual (self):
        if self._cambio_automatico:
            self._cambio_automatico = True
            print(f"{self.modelo} alterou para o manual")
        else:
            print(f"{self.modelo} já está no manual")

    #Alterar cambio para automático
    def alterar_para_automatico(self):
        if self._cambio_automatico:
            self._cambio_automatico = False
            print(f"{self.modelo} alterou para o automático")
        else:
            print(f"{self.modelo} já está no automático")

    #Alterar cor
    def alterar_cor(self):
        if self.cor == "Branca":
            cor_nova = input("Para qual cor você deseja alterar: ")
            self.cor = cor_nova
            print(f"Agora o {self.modelo} está na cor {self.cor}")

    #Alterar tração

    def alterar_tracao(self):
        if self.tp_tracao == 2:
            self.tp_tracao = 4
        
        else:
            self.tp_tracao = 2
        return True

    def mudar_potencia (self, pot="1.0"):
        try:
            v1_pot=float(pot)
            self.potencia=v1_pot
            return True
        except:
            return False


       
```

```python
meu_carro = Carro ("Vollswagem", "voyage")
meu_carro.mudar_potencia(2.0)
```

```python
print(f"A nova potência do {meu_carro.modelo} é: {meu_carro.potencia}")
```
