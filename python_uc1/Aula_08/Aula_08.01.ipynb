```python
def potencia (numero, expoente):
    resultado = numero ** expoente
    return resultado
```

```python
n1 = 2
exp = 8

print(potencia (expoente=exp, numero=n1))
```

```python
n2=5
exp=4
resultado = potencia (n2, exp)
resultado2 = potencia(exp, n2)
```

```python
print(f"{resultado} <> {resultado2}")
```

### Atividade 4

```python
def fatorial (n):
    if n < 0:
        return "Número inválido"
    elif n == 0:
        return 1
    else:
        resultado = 1 
        for i in range (1, n + 1):
            resultado *= i
        return resultado
```

```python
print("O fatorial de 7 é: ", fatorial (7))
```

### Atividade 7

```python
def eh_palidromo (palavra):
    return palavra == palavra [::-1]
```

```python
print("Ana é um palidromo?", eh_palidromo("ana"))
print("Flamengo é palidromo?", eh_palidromo("Flamengo"))
```

### Atividade 8

```python
def contar_vogais(texto):
    vogais = ("AEIOUaeiou")
    contador = 0
    for char in texto:
        if char in vogais:
            contador += 1
    return contador
```

```python
print("Número de vogais em Flamengo", contar_vogais("Flamengo"))
```

### Atividade 9

```python
def tabuada(n):
    resultado = []
    for i in range (1, 11):
        resultado.append (n * i)
    return resultado
```

```python
print("A tabuada de 9 é", tabuada(9))
```

### Atividade 10

```python
def ordenar_por_segundo(lista_tuplas):
    return sorted (lista_tuplas, key=lambda x: x[1])
```

```python
lista_exemplo = [("a", 1), ("b", 2), ("c", 3)]
print("Lista ordenada:", ordenar_por_segundo(lista_exemplo))
```
