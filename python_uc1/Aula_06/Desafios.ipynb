1️⃣ Verificação de Acesso ao Sistema
O usuário só pode acessar se tiver mais de 18 anos e uma senha correta

```python
senha=2020

idade = int(input("Digite a sua idade: "))
senha_digitada=int(input("Digite a senha: "))

if idade >=18 and senha_digitada==senha:
    print("Você está autorizado a entrar no sistema")
else:
    print("Você não está autorizado a entrar no sistema")
```

2️⃣ Aprovado na Faculdade?
O aluno passa se a média for ≥ 7 ou se tiver feito um trabalho extra.

```python
n1=float(input("Digite a PRIMEIRA NOTA"))
n2=float(input("Digite a SEGUNDA NOTA"))
n3=float(input("Digite a TERCEIRA NOTA"))
n4=float(input("Digite a QUARTA NOTA"))
media=(n1+n2+n3+n4)/4
resp_extra=input("Fez o trabalho extra?")
if media>=7 or resp_extra=="sim":
    print("Você está aprovado")
else:
    print("Você está reprovado")
```

3️⃣ Controle de Entrada em uma Festa
Para entrar, a pessoa precisa ter:

✅ Mais de 18 anos OU estar acompanhada dos pais.-

✅ Não estar na lista de banidos.

```python
idade =int(input("Digite a idade: "))
acomp_resp=input("Você está acompanhado")
banido=input("Consta na lista de banidos")

if (idade >= 18 or acomp_resp == "sim") and banido == "nao":
        print("Autorizado a entrar na festa")
else:
    print("Você não está autorizado a entrar na festa")
```
