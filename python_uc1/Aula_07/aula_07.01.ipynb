##  Imprimir números

## Quadrados dos números

```python
for numero in range (1, 11):
    quadrado=numero**2
    print (f"O quadrado de .: {numero} :. é .:{quadrado}")
```

## Pares e Ímpares

```python
print(f"Os números pares de 1 a 20 são:")
for numero in range (1, 21):
    if numero % 2 == 0:
        print (numero, end=", ")
```

```python
print(f"Os números ímpares de 1 a 20 são:")
for numero in range (1, 21):
    if numero % 2 != 0:
        print(numero, end=", ")
```

```python
pares=[]
impares=[]
for numero in range (1, 21):
    if (numero % 2 == 0):
        pares.append(numero)
    else:
        impares.append(numero)

print (f"Números pares   : {pares}")
print (f"Números ímpares : {impares}")
```

## Contagem regressiva

```python
numero = 10
print("Contagem regressiva: ")

while numero >= 0:
    print(numero, end= ", ")
    numero -= 1
```

## Programa para perguntar o nome

```python
nome = input ("Digite seu nome: ")
while (nome != 'Sair'):
    nome = input ("digite novamente seu nome: ")
    if (nome == 'Sair'):
            print("Você saiu.")
                
```

## Soma dos números de 1 a 100

```python
for numero in range (1, 101):
    soma = numero + numero
    print(f"A soma de {numero} + {numero} é:", soma )
```

```python
soma = 0
for numero in range (1, 101):
    soma = soma + numero
print (f"O valor da soma de todos os números de 1 até 100 é {soma}")
```

#### usando while

```python
soma = 0
numero = 1
while numero <= 100 :
    soma = soma + numero
    numero = numero + 1 
print (f"A soma de todos os números é: {soma}")
```

## Fatorial de um número

```python
numero = int(input("Digite um número: "))

fatorial = 1

while numero > 1:
    fatorial *= numero
    numero -= 1
print ("Fatorial: ", fatorial)
    
```
