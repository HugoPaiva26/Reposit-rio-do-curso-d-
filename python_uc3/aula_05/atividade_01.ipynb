# Formas de armazenamentos de dados

## Estruturas em memórias

### 1 - Listas (list)

* O que é:

Uma lista é uma coleção ordenada de itens, criada com colchetes e separada por vírgulas. Ela é útil para agrupar valores relacionados em uma única variável.


```python
# Lista de produtos
produtos = ["arroz", "feijão", "leite"]
```

* Quando utilizar: 


Use listas quando a ordem dos itens importa e o conjunto pode mudar ao longo do programa, como produtos de um pedido, tarefas ou nomes de participantes. Em Django, elas são úteis para organizar dados temporários em memória; dados que precisam ser salvos normalmente ficam em modelos e no banco de dados.

Características:
- Ordenada: os itens têm posições, chamadas índices, começando em 0.
- Mutável: é possível alterar, adicionar e remover itens.
- Aceita repetições: o mesmo valor pode aparecer mais de uma vez.
- Aceita tipos diferentes: por exemplo, textos e números, embora itens do mesmo tipo sejam mais fáceis de trabalhar.

Exemplo principais métodos:

```python
# Lista inicial de produtos
produtos = ["Arroz", "Feijão", "Leite"]

# append(): adiciona um elemento ao final
produtos.append("Café")

# insert(): adiciona um elemento na posição indicada
produtos.insert(1, "Açúcar")

# extend(): adiciona vários elementos
produtos.extend(["Macarrão", "Óleo"])

# remove(): remove a primeira ocorrência do valor
produtos.remove("Leite")

# pop(): remove e retorna o elemento da posição indicada
removido = produtos.pop(0)

# index(): retorna a posição do elemento
posicao = produtos.index("Café")
```

### 2 - Tupla (tuple)

* Uma tupla é uma estrutura de dados utilizada para armazenar vários valores em uma única variável.
* Diferentemente das listas, tuplas são imutáveis, ou seja, não permitem adicionar, remover ou alterar elementos após sua criação.

```python
produto = ("Arroz", 25.90, 10)
# Nome, preço e quantidade em estoque
```

##### Características
- Ordenada: mantém a ordem dos elementos.
- Imutável: não permite alterar, adicionar ou remover elementos.
- Indexada: permite acessar elementos pela posição.
- Aceita duplicados: pode armazenar valores repetidos.
- Heterogênea: aceita diferentes tipos de dados.
- Permite aninhamento: pode conter outras tuplas ou listas

##### Quando utilizar?
- Utilize tuplas quando precisar armazenar dados que não devem ser alterados durante a execução do programa.
- Exemplos: coordenadas geográficas, dias da semana, configurações fixas e informações de produtos.

##### Principais métodos — Exemplo prático
- Imagine um supermercado que registra as categorias dos produtos vendidos.
- As tuplas possuem apenas dois métodos próprios: count() e index().

```python
# Tupla com categorias dos produtos
categorias = ("Alimentos", "Bebidas", "Limpeza",
              "Alimentos", "Higiene")

# count(): conta quantas vezes um valor aparece
quantidade = categorias.count("Alimentos")
print(quantidade)  # 2

# index(): retorna a posição da primeira ocorrência
posicao = categorias.index("Limpeza")
```

##### Operações úteis com tuplas

```python
produtos = ("Arroz", "Feijão", "Leite")

# Acessa o primeiro elemento
print(produtos[0])  # Arroz

# Acessa o último elemento
print(produtos[-1])  # Leite

# Retorna a quantidade de elementos
print(len(produtos))  # 3

# Verifica se um elemento existe
print("Arroz" in produtos)  # True

# Percorre todos os elementos
for produto in produtos:
    print(produto)

# Desempacota os elementos em variáveis
nome1, nome2, nome3 = produtos
print(nome1)  # Arroz
```

### 3- Dicionário (dict)

##### O que é?
* Um dicionário é uma estrutura de dados que armazena informações em pares de chave e valor.
* Cada chave identifica um valor, permitindo acessar informações diretamente.

```python
produto = {
    "nome": "Arroz",
    "preco": 25.90,
    "estoque": 10
}
```

##### Características
- Mutável: permite adicionar, alterar e remover elementos.
- Ordenado: mantém a ordem de inserção dos elementos.
- Chaves únicas: não permite chaves duplicadas.
- Heterogêneo: aceita diferentes tipos de valores.
- Acesso por chave: utiliza chaves em vez de índices numéricos.
- Chaves hashable: aceita chaves como strings, números e tuplas de elementos imutáveis.

##### Quando utilizar?
* Utilize dicionários quando precisar organizar informações relacionadas e acessá-las por meio de uma chave.
* Exemplos: cadastro de produtos, informações de clientes, configurações e dados de formulários.

Principais métodos — Exemplo prático

```python
# Dicionário inicial
produto = {
    "nome": "Arroz",
    "preco": 25.90,
    "estoque": 10
}

# get(): consulta um valor pela chave
nome = produto.get("nome")

# keys(): retorna as chaves
chaves = produto.keys()

# values(): retorna os valores
valores = produto.values()

# items(): retorna os pares de chave e valor
itens = produto.items()

# update(): atualiza ou adiciona informações
produto.update({"preco": 27.90})

# setdefault(): retorna o valor ou cria a chave
categoria = produto.setdefault("categoria", "Alimentos")

# pop(): remove uma chave e retorna seu valor
estoque = produto.pop("estoque")

# popitem(): remove e retorna o último par inserido
ultimo = produto.popitem()

# copy(): cria uma cópia superficial
copia = produto.copy()

# clear(): remove todos os elementos
produto.clear()
```

### 4 - Conjunto (set)

##### O que é?
* Um conjunto é uma estrutura de dados utilizada para armazenar elementos únicos, sem permitir duplicações.

```python
categorias = {"Alimentos", "Bebidas", "Limpeza"}
```

##### Características
- Mutável: permite adicionar e remover elementos.
- Não ordenado: não garante a ordem dos elementos.
- Sem duplicados: armazena apenas valores únicos.
- Não indexado: não permite acessar elementos por posição.
- Elementos hashable: aceita números, strings e tuplas de elementos imutáveis, mas não listas ou dicionários comuns.
- Operações matemáticas: permite união, interseção e diferença.

##### Quando utilizar?
* Utilize conjuntos quando precisar eliminar duplicações, verificar a existência de elementos ou comparar coleções.
* Exemplos: categorias de produtos, códigos únicos, clientes cadastrados e comparação de estoques.

Exemplo prático - principais métodos:

```python
# Conjunto inicial
categorias = {"Alimentos", "Bebidas", "Limpeza"}

# add(): adiciona um elemento
categorias.add("Higiene")

# update(): adiciona vários elementos
categorias.update(["Padaria", "Congelados"])

# remove(): remove um elemento existente
categorias.remove("Bebidas")

# discard(): remove um elemento, se existir
categorias.discard("Eletrônicos")

# copy(): cria uma cópia independente
copia = categorias.copy()

# Conjunto de categorias de outra filial
filial = {"Alimentos", "Higiene", "Pet Shop"}

# union(): reúne elementos dos dois conjuntos
uniao = categorias.union(filial)

# intersection(): retorna elementos em comum
comuns = categorias.intersection(filial)

# difference(): retorna elementos exclusivos do primeiro
diferenca = categorias.difference(filial)

# symmetric_difference(): retorna elementos não compartilhados
exclusivos = categorias.symmetric_difference(filial)

# issubset(): verifica se um conjunto está contido em outro
subconjunto = {"Alimentos"}.issubset(categorias)

# issuperset(): verifica se contém todos os elementos
superconjunto = categorias.issuperset({"Higiene"})

# isdisjoint(): verifica se não existem elementos em comum
separados = categorias.isdisjoint({"Eletrônicos"})

# pop(): remove e retorna um elemento arbitrário
removido = categorias.pop()

# clear(): remove todos os elementos
categorias.clear()
```

## Persistência de dados

#### O que é persistência de dados?
Persistência de dados é a capacidade de armazenar informações de forma que elas continuem disponíveis mesmo depois que o programa é encerrado.
Por exemplo, se você armazena um nome em uma variável, esse dado normalmente deixa de estar disponível quando o programa termina. Se você salva esse nome em um arquivo, pode recuperá-lo quando executar o programa novamente.
Existem diferentes formas de persistir dados em Python.

## 1 - Arquivos Simples

### A - Arquivos de texto (.txt)

##### O que são
* Arquivos que armazenam informações e formato de texto simples
##### Quando usar:
* Para salvar anotações, nomes, resultados, configurações simples ou registros que podem ser lidos como texto.

##### Características
- Fácil de criar e ler.
- Armazena textos.
- Permite leitura e escrita.
- Não possui estrutura de dados definida.

Exemplo pratico - principais métodos

```python
# Cria o arquivo e grava um produto
with open("produtos.txt", "w", encoding="utf-8") as arquivo:
    arquivo.write("Arroz\n")

# Adiciona outro produto sem apagar o anterior
with open("produtos.txt", "a", encoding="utf-8") as arquivo:
    arquivo.write("Feijão\n")

# read(): lê todo o conteúdo
with open("produtos.txt", "r", encoding="utf-8") as arquivo:
    print(arquivo.read())

# readline(): lê a primeira linha
with open("produtos.txt", "r", encoding="utf-8") as arquivo:
    print(arquivo.readline().strip())  # Arroz

# readlines(): retorna as linhas em uma lista
with open("produtos.txt", "r", encoding="utf-8") as arquivo:
    print(arquivo.readlines())  # ['Arroz\n', 'Feijão\n']
```

### B - Arquivos CSV (.csv)

##### O que é?
* CSV (Comma-Separated Values) é um formato utilizado para armazenar dados organizados em linhas e colunas, separados por delimitadores.

##### Características
- Estrutura tabular.
- Compatível com Excel e outras planilhas.
- Utiliza separadores como vírgula ou ponto e vírgula.
- Armazena os valores como texto.

##### Quando utilizar?
* Para armazenar tabelas, relatórios de vendas, cadastros e dados que serão utilizados em planilhas.

Exemplo

```python
import csv

# writer() e writerow(): criam o arquivo e gravam linhas
with open("produtos.csv", "w", newline="", encoding="utf-8") as arquivo:
    escritor = csv.writer(arquivo)
    escritor.writerow(["Nome", "Preço", "Estoque"])
    escritor.writerow(["Arroz", 25.90, 10])

# writerows(): adiciona várias linhas
with open("produtos.csv", "a", newline="", encoding="utf-8") as arquivo:
    escritor = csv.writer(arquivo)
    escritor.writerows([
        ["Feijão", 8.50, 20],
        ["Leite", 5.90, 15]
    ])

# reader(): lê os dados como listas
with open("produtos.csv", "r", newline="", encoding="utf-8") as arquivo:
    leitor = csv.reader(arquivo)
    for linha in leitor:
        print(linha)

# DictReader(): lê cada linha como dicionário
with open("produtos.csv", "r", newline="", encoding="utf-8") as arquivo:
    leitor = csv.DictReader(arquivo)
    for produto in leitor:
        print(produto["Nome"])

# DictWriter(): grava dados utilizando chaves
with open("copia.csv", "w", newline="", encoding="utf-8") as arquivo:
    escritor = csv.DictWriter(arquivo, fieldnames=["Nome", "Preço", "Estoque"])
    escritor.writeheader()
    escritor.writerow({"Nome": "Arroz", "Preço": 25.90, "Estoque": 10})
```

## 2 - Serialização de Objetos (JSON e Pickle)

Serialização é o processo de transformar um dado ou objeto em um formato que possa ser salvo em um arquivo ou transmitido para outro sistema, permitindo recuperá-lo posteriormente.

### A - Arquivos JSON (.json)

##### O que é?
* JSON (JavaScript Object Notation) é um formato textual utilizado para armazenar dados estruturados, como listas e dicionários.

##### Características
- Fácil de ler e interpretar.
- Organiza dados em estruturas de chave e valor.
- Compatível com diferentes linguagens.
- Muito utilizado em APIs e configurações.

##### Quando utilizar?
* Para salvar cadastros, configurações, informações estruturadas e dados trocados entre sistemas.

Exemplo

```python
import json

# Dados do produto
produto = {
    "nome": "Arroz",
    "preco": 25.90,
    "estoque": 10
}

# dump(): salva os dados em um arquivo JSON
with open("produto.json", "w", encoding="utf-8") as arquivo:
    json.dump(produto, arquivo, indent=4, ensure_ascii=False)

# load(): recupera os dados do arquivo
with open("produto.json", "r", encoding="utf-8") as arquivo:
    dados = json.load(arquivo)
    print(dados["nome"])  # Arroz

# dumps(): converte o dicionário em string JSON
texto = json.dumps(produto, ensure_ascii=False)
print(texto)

# loads(): converte a string JSON em dicionário
novo_produto = json.loads(texto)
print(novo_produto["estoque"])  # 10
```

### B - Pickle (.pkl)

##### O que é?
* Pickle é um módulo Python utilizado para serializar e desserializar objetos Python, permitindo salvá-los em arquivos binários.

* Características
- Armazena objetos Python.
- Utiliza formato binário.
- Preserva diversos tipos de objetos Python.
- Não é indicado para troca de dados com outras linguagens.
- Não é seguro carregar arquivos de fontes desconhecidas.

##### Quando utilizar?
* Para salvar objetos Python em aplicações locais, quando os arquivos são gerados e controlados pela própria aplicação.

Exemplo

```python
import pickle

# Dados do produto
produto = {
    "nome": "Arroz",
    "preco": 25.90,
    "estoque": 10
}

# dump(): salva o objeto em arquivo binário
with open("produto.pkl", "wb") as arquivo:
    pickle.dump(produto, arquivo)

# load(): recupera o objeto do arquivo
with open("produto.pkl", "rb") as arquivo:
    dados = pickle.load(arquivo)
    print(dados["nome"])  # Arroz

# dumps(): transforma o objeto em bytes
dados_binarios = pickle.dumps(produto)

# loads(): reconstrói o objeto a partir dos bytes
novo_produto = pickle.loads(dados_binarios)
print(novo_produto["estoque"])  # 10
```

##  Bancos de Dados (SQLite)

##### O que é?
* SQLite é um banco de dados relacional que armazena informações em um arquivo, permitindo realizar operações de cadastro, consulta, atualização e exclusão.

##### Características
- Não necessita de servidor separado.
- Armazena dados em tabelas.
- Utiliza comandos SQL.
- Suporta transações.
- Possui integração nativa com Python por meio de sqlite3.

##### Quando utilizar?
* Para sistemas de cadastro, controle de estoque, aplicações locais, protótipos e aplicações web de pequeno porte.

Exemplo:

```python
import sqlite3

# connect(): conecta ou cria o banco de dados
conexao = sqlite3.connect("supermercado.db")

# cursor(): cria um cursor para executar comandos
cursor = conexao.cursor()

# execute(): cria a tabela, caso não exista
cursor.execute("""
    CREATE TABLE IF NOT EXISTS produtos (
        id INTEGER PRIMARY KEY,
        nome TEXT,
        preco REAL,
        estoque INTEGER
    )
""")

# execute(): cadastra um produto
cursor.execute(
    "INSERT INTO produtos (nome, preco, estoque) VALUES (?, ?, ?)",
    ("Arroz", 25.90, 10)
)

# Guarda o ID do produto cadastrado
produto_id = cursor.lastrowid

# executemany(): cadastra vários produtos
cursor.executemany(
    "INSERT INTO produtos (nome, preco, estoque) VALUES (?, ?, ?)",
    [
        ("Feijão", 8.50, 20),
        ("Leite", 5.90, 15)
    ]
)

# commit(): confirma as alterações
conexao.commit()

# fetchone(): consulta um único registro
cursor.execute("SELECT * FROM produtos WHERE id = ?", (produto_id,))
print(cursor.fetchone())

# fetchall(): consulta todos os registros
cursor.execute("SELECT * FROM produtos")
print(cursor.fetchall())

# execute(): atualiza o preço do produto
cursor.execute(
    "UPDATE produtos SET preco = ? WHERE id = ?",
    (27.90, produto_id)
)

# execute(): exclui o produto cadastrado
cursor.execute("DELETE FROM produtos WHERE id = ?", (produto_id,))

# commit(): confirma a atualização e exclusão
conexao.commit()

# close(): fecha a conexão
conexao.close()
```
